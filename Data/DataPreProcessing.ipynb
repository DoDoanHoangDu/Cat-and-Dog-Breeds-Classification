```python
import os
import shutil
import random
```

```python
def get_subfolder_names(folder_path):
    return [name for name in os.listdir(folder_path) if os.path.isdir(os.path.join(folder_path, name))]
```

```python
def delete_folder(folder_path):
    if os.path.isdir(folder_path):
        shutil.rmtree(folder_path)
        print(f"Folder '{folder_path}' deleted successfully.")
    else:
        print(f"Folder '{folder_path}' does not exist or is not a directory.")
```

```python
def split_subfolders(source_folder, training_set_folder, validation_set_folder, k=0.8,target_subfolders = None,delete = True):
    #Delete old folder to ensure no duplication/fresh start
    if delete:
        delete_folder(training_set_folder)
        delete_folder(validation_set_folder)
    if not (0 <= k <= 1):
        raise ValueError("k must be between 0 and 1")
    os.makedirs(training_set_folder, exist_ok=True)
    os.makedirs(validation_set_folder, exist_ok=True)
    subfolders = target_subfolders if target_subfolders is not None else os.listdir(source_folder)
    for subfolder in subfolders:
        subfolder_path = os.path.join(source_folder, subfolder)
        if os.path.isdir(subfolder_path):
            print(f"Processing subfolder: {subfolder}")
            training_subfolder_path = os.path.join(training_set_folder, subfolder)
            validation_subfolder_path = os.path.join(validation_set_folder, subfolder)
            os.makedirs(training_subfolder_path, exist_ok=True)
            os.makedirs(validation_subfolder_path, exist_ok=True)
            files = [f for f in os.listdir(subfolder_path) if os.path.isfile(os.path.join(subfolder_path, f))]
            random.shuffle(files)
            split_index = int(len(files) * k)
            training_files = files[:split_index]
            validation_files = files[split_index:]
            for file_name in training_files:
                shutil.copy(os.path.join(subfolder_path, file_name), os.path.join(training_subfolder_path, file_name))
            for file_name in validation_files:
                shutil.copy(os.path.join(subfolder_path, file_name), os.path.join(validation_subfolder_path, file_name))
            print(f"Moved {len(training_files)} files to {training_subfolder_path} and {len(validation_files)} files to {validation_subfolder_path}")
```

```python
def copy__files(source_folder, destination_folder, k):
    if not (0 <= k <= 1):
        raise ValueError("k must be between 0 and 1")
    os.makedirs(destination_folder, exist_ok=True)
    files = [f for f in os.listdir(source_folder) if os.path.isfile(os.path.join(source_folder, f))]
    num_files_to_copy = int(len(files) * k)
    selected_files = random.sample(files, num_files_to_copy)
    for file_name in selected_files:
        shutil.copy(os.path.join(source_folder, file_name), os.path.join(destination_folder, file_name))
    print(f"Copied {num_files_to_copy} files from '{source_folder}' to '{destination_folder}'.")
```

```python
def count_images_in_folder(folder_path):
    image_extensions = ['.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp']
    image_count = 0
    for dirpath, _, filenames in os.walk(folder_path):
        for file_name in filenames:
            _, ext = os.path.splitext(file_name)
            if ext.lower() in image_extensions:
                image_count += 1
    print(f"Number of Images in {folder_path}:",image_count)
    return image_count
```

```python
split_ratio = 0.8

original_cats = "CatBreedsOriginal"
cat_breeds_train = r"CatBreeds\train"
cat_breeds_test = r"CatBreeds\test"

original_dogs = "DogBreedsOriginal"
dog_breeds_train = r"DogBreeds\train"
dog_breeds_test = r"DogBreeds\test"

original_cd = "CatDogOriginal"
original_cd_cats_train = r"CatDogOriginal\training_set\training_set\cats"
original_cd_dogs_train = r"CatDogOriginal\training_set\training_set\dogs"

original_cd_cats_test = r"CatDogOriginal\test_set\test_set\cats"
original_cd_dogs_test = r"CatDogOriginal\test_set\test_set\dogs"
```

```python
cd = "CatDog"
delete_folder(cd)
shutil.copytree(original_cd, cd)
cd_cats_train = r"CatDog\training_set\training_set\cats"
cd_dogs_train = r"CatDog\training_set\training_set\dogs"

cd_cats_test = r"CatDog\test_set\test_set\cats"
cd_dogs_test = r"CatDog\test_set\test_set\dogs"
```

```python
split_subfolders(original_cats, cat_breeds_train, cat_breeds_test, split_ratio)
```

```python
split_subfolders(original_dogs, dog_breeds_train, dog_breeds_test, split_ratio)
```

```python
cat_train_ratio = min(1, count_images_in_folder(dog_breeds_train)/count_images_in_folder(cat_breeds_train))
cat_test_ratio = min(1, count_images_in_folder(dog_breeds_test)/count_images_in_folder(cat_breeds_test))
dog_train_ratio = min(1, count_images_in_folder(cat_breeds_train)/count_images_in_folder(dog_breeds_train))
dog_test_ratio = min(1, count_images_in_folder(cat_breeds_test)/count_images_in_folder(dog_breeds_test))
```

```python
all_cats_names = get_subfolder_names(original_cats)
all_dogs_name = get_subfolder_names(original_dogs)
print(all_cats_names)
print(all_dogs_name)
```

```python
for name in all_cats_names:
    copy__files(source_folder=os.path.join(cat_breeds_train,name),
                destination_folder=cd_cats_train,
                k = cat_train_ratio)

    copy__files(source_folder=os.path.join(cat_breeds_test,name),
                destination_folder=cd_cats_test,
                k = cat_test_ratio)
for name in all_dogs_name:
    copy__files(source_folder=os.path.join(dog_breeds_train,name),
                destination_folder=cd_dogs_train,
                k = dog_train_ratio)

    copy__files(source_folder=os.path.join(dog_breeds_test,name),
                destination_folder=cd_dogs_test,
                k = dog_test_ratio)
```

```python
count_images_in_folder(cd_cats_train)
count_images_in_folder(cd_dogs_train)
count_images_in_folder(cd_cats_test)
count_images_in_folder(cd_dogs_test)
```
